import numpy as np
from IPython.display import Audio, display

def pitido(frec=440, segs=0.5):
    instantes = np.linspace(0, segs, int(22050 * segs))
    sonido = np.sin(2 * np.pi * frec * instantes)
    display(Audio(sonido, rate=22050, autoplay=True))
